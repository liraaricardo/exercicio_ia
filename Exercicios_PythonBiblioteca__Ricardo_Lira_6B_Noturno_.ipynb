4. Gere um array de 100 numeros inteiros aleatorios e encontre a mediana e
a moda dos valores.

```python
import numpy as np
from scipy import stats

array = np.random.randint(1, 101, size=100)

mediana = np.median(array)
moda = stats.mode(array, keepdims=True)

print("Array:", array)
print("Mediana:", mediana)
print("Moda:", moda.mode[0], "- Frequência:", moda.count[0])
```

Explicação: NumPy não tem função nativa de moda, por isso é usado scipy.stats.mode. np.median() calcula a mediana normalmente.

9. Crie um array com 10 numeros aleatorios e normalize seus valores entre 0
e 1.

```python
import numpy as np

array = np.random.randint(1, 100, size=10)
print("Original:", array)

normalizado = (array - np.min(array)) / (np.max(array) - np.min(array))
print("Normalizado:", normalizado)
```

19. Crie dois arrays de numeros inteiros aleatorios de tamanho 20 e calcule o
angulo entre eles (produto escalar normalizado)

```python
import numpy as np

array1 = np.random.randint(1, 20, size=20)
array2 = np.random.randint(1, 20, size=20)

produto_escalar = np.dot(array1, array2)
norma1 = np.linalg.norm(array1)
norma2 = np.linalg.norm(array2)

cos_theta = produto_escalar / (norma1 * norma2)
angulo_rad = np.arccos(cos_theta)
angulo_graus = np.degrees(angulo_rad)

print("Array 1:", array1)
print("Array 2:", array2)
print("Cosseno do ângulo:", cos_theta)
print("Ângulo (graus):", angulo_graus)
```

Explicação: O cosseno do ângulo entre dois vetores é dado pelo produto escalar dividido pelo produto das normas. np.arccos() converte o cosseno em radianos, e np.degrees() converte para graus.

1. Calcule a integral da funcao f(x) = x2 sin(x) no intervalo de 0 a π.

```python
from scipy import integrate
import numpy as np

def f(x):
    return x**2 * np.sin(x)

resultado, erro = integrate.quad(f, 0, np.pi)
print("Resultado da integral:", resultado)
print("Erro estimado:", erro)
```

Explicação: integrate.quad() calcula a integral numérica definida no intervalo especificado, retornando o resultado e uma estimativa do erro.

2. Resolva a equacao diferencial dy
dx = x−y, com y(0) = 1, no intervalo de 0
a 10.

```python
from scipy.integrate import solve_ivp
import numpy as np

def eq_diferencial(x, y):
    return x - y

solucao = solve_ivp(eq_diferencial, [0, 10], [1], t_eval=np.linspace(0, 10, 100))

print("x:", solucao.t)
print("y:", solucao.y[0])
```

Explicação: solve_ivp resolve a EDO numericamente. t_eval define os pontos onde queremos avaliar a solução, gerando uma curva suave.

3. Encontre as raızes da funcao f(x) = x4 − 3x3 + 2 utilizando SciPy

```python
from scipy.optimize import fsolve
import numpy as np

def f(x):
    return x**4 - 3*x**3 + 2

estimativas = [-1, 0.5, 1, 3]
raizes = [fsolve(f, x0)[0] for x0 in estimativas]

print("Raízes encontradas:", raizes)
```

Explicação: Como fsolve encontra uma raiz por vez a partir de um ponto inicial, usamos várias estimativas para tentar localizar diferentes raízes da equação (função de grau 4 pode ter até 4 raízes reais).

4. Use a funcao de otimizacao de SciPy para maximizar f(x) = −x2+4x+1.

```python
from scipy.optimize import minimize

def f_negativa(x):
    return -(-x**2 + 4*x + 1)

resultado = minimize(f_negativa, 0)
x_max = resultado.x[0]
valor_max = -resultado.fun

print("x que maximiza:", x_max)
print("Valor máximo:", valor_max)
```

7. Use a funcao de interpolacao cubica de SciPy para interpolar os dados
(0, 0), (1,2),(2,3),(3,5),(4,4).

```python
from scipy.interpolate import interp1d
import numpy as np

x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 2, 3, 5, 4])

f_cubica = interp1d(x, y, kind='cubic')

x_novos = np.linspace(0, 4, 20)
y_novos = f_cubica(x_novos)

print("Valores interpolados:", y_novos)
```

9. Ajuste uma curva quadratica aos pontos (1,2),(2,4),(3,5),(4,7),(5,8).

```python
from scipy.optimize import curve_fit
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])

def quadratica(x, a, b, c):
    return a * x**2 + b * x + c

parametros, _ = curve_fit(quadratica, x, y)
print("Parâmetros [a, b, c]:", parametros)
```

10. Calcule a integral dupla da funcao f(x,y) = xy + x2 no intervalo [0,2]
para x e [0,1] para y.

```python
from scipy import integrate

def f(y, x):
    return x*y + x**2

resultado, erro = integrate.dblquad(f, 0, 2, 0, 1)

print("Resultado da integral dupla:", resultado)
print("Erro estimado:", erro)
```
