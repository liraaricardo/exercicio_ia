2.Soma de Dois Números
o Crie um programa que peça ao usuário dois números e exiba a soma
deles.

```python
#variavel para o usuario digitar os numeros
n1=float(input("Digite o primeiro numero :"));
n2=float(input("Digite o segundo numero :"));
#declarando variavel da soma
soma= n1+n2;
#exibindo os valores utilizando concatenação
print(f"{n1} + {n2} = {soma}")
```

3.Cálculo da Área do Círculo
o Solicite o raio de um círculo e calcule sua área. Use a fórmula área =
π * r².

```python



#variavel para o usuario digitar o valor do raio
raio=float(input("Digite o valor do raio do circulo: "));
pi=3.14
#calculo da area utilizando essa formula π * r².
area=pi*(raio**2);
#exibindo os valores utilizando concatenação
print(f"a area do circulo é: {area}")
```

4.Conversão de Temperatura
o Converta uma temperatura fornecida em Celsius para Fahrenheit
usando a fórmula F = C * 9/5 + 32.

```python
#criando variavel para o usuario digitar a temperatura
celsius=float(input("Digite a temperatura em celcius: "));

#calculo da conversão da temperatura
fahrenheit= celsius * 9/5 + 32 ;

print(f"{celsius}º Equivale a {fahrenheit}ºF ");
```

8.Verificação de Palíndromo
o Crie um programa que verifique se uma palavra é um palíndromo
(ex: "radar").

```python
#variavel para o usuario digitar uma palavra
palavra = input("Digite uma palavra: ");
if palavra == palavra[::-1]:#condiocinal utilizando [::-1] para comparar a palavra original com ela invertida
  print("A palavra é um palíndromo")
else:
  print(f"A palavra {palavra} não é um palíndromo");
```

9.Fatorial de um Número
o Calcule o fatorial de um número fornecido pelo usuário.

```python
numero = int(input("Digite um número: "))#variavel para o usuario digitar um numero
fatorial = 1
for i in range(1, numero + 1):#laço de repetição para calcular o fatorial
    fatorial *= i

print(f"O fatorial de {numero} é {fatorial}")
```

11.Lista de Números
o Crie uma lista de números e exiba o maior e o menor número.

```python
#criação da Lista
numeros=[10,50,40,30,20,15,23,12];

maior=max(numeros);#função para saber o maior numero
menor=min(numeros);#função para saber o menor

print(f"Lista: {numeros}");

print(f"Maior: {maior}");

print(f"Menor: {menor}");
```
