12.Contagem de Caracteres o Conte o número de ocorrências de cada caractere em uma string
fornecida.

```python
texto=input("Digite uma palavra: ");
contagem ={}#cria um dicionário vazio para guardar cada caractere e sua quantidade

#para cada caractere na string verifica se já esta no dicionario se estiver incrementa a contagem (+= 1).
#se não adiciona ao dicionário com valor inicial 1
for caractere in texto:
  if caractere in contagem:
    contagem[caractere]+=1
  else:
    contagem[caractere]=1
for caractere,quantidade in contagem.items():#percorre o dicionário retornando pares (chave, valor), que são impressos
  print(f"'{caractere}' : {quantidade} ")
```

17. Dicionário de Contagem de Palavras
o Conte a frequência de cada palavra em uma string fornecida pelo
usuário.

```python
texto= input("digite uma frase: ")
palavras=texto.split()#divide a string em uma lista de palavras separando pelos espaços em branco

contagem= {}

for palavra in palavras:
  if palavra in contagem:
    contagem[palavra]+=1
  else:
    contagem[palavra]=1

for palavra,quantidade in contagem.items():
  print(f"'{palavra}': {quantidade}")

```

18. Número Primo
o Crie um programa que verifique se um número fornecido é primo.

```python
numero=int(input("Digite um numero: "))
primo=True

if numero<2:
  primo=False
else:
  for i in range(2,int(numero**0.5)+1): #o loop testa se numero é divisível por algum valor de 2 até a raiz quadrada dele
    if numero % i==0: #Se encontrar algum divisor  marca primo = False e usa break para sair do loop
      primo=False
      break

if primo:
  print(f"{numero} é primo")
else:
   print(f"{numero} não é primo")

```

19. Jogo de Adivinhação
o Implemente um jogo onde o usuário deve adivinhar um número
gerado aleatoriamente pelo computador.

```python
import random

numero_secreto = random.randint(1, 100)#gera um número inteiro aleatório entre 1 e 100

print("Adivinhe o número entre 1 e 100!")

while True: #cria um loop que só é interrompido pelo break quando o usuário acerta
    palpite = int(input("Digite seu palpite: "))

    if palpite == numero_secreto:
        print("Parabéns! Você acertou!")
        break
    elif palpite < numero_secreto:
        print("Muito baixo! Tente novamente.")
    else:
        print("Muito alto! Tente novamente.")
```

20. Matriz Transposta
o Dada uma matriz (lista de listas), crie sua matriz transposta.

```python
matriz=[
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
transposta=[]

for coluna in range(len(matriz[0])):#o loop externo (coluna) percorre cada coluna da matriz original
  nova_linha=[]
  for linha in range(len(matriz)):#o loop interno (linha) percorre cada linha pegando o elemento matriz[linha]
    nova_linha.append(matriz[linha][coluna])
  transposta.append(nova_linha)

print("Matriz original:")
for linha in matriz:
    print(linha)

print("\nMatriz transposta:")
for linha in transposta:
    print(linha)
```

21. Busca Binária
o Implemente a busca binária para encontrar um elemento em uma
lista ordenada.

```python
def busca_binaria(lista,alvo):
  inicio=0
  fim=len(lista)-1 #inicio e fim marcam os limites da parte da lista que ainda estamos considerando

  while inicio<=fim:
    meio=(inicio+fim)//2 #calcula o índice do meio

    if lista[meio]==alvo: #achou,retorna a posição
      return meio

    elif lista[meio]<alvo:#o valor procurado está na metade direita, então move o inicio para depois do meio
      inicio=meio+1
    else:
      fim=meio-1
#o loop continua até inicio > fim (não sobrou mais nada pra buscar) ou até encontrar o valor,se não encontrar nada, retorna -1
  return -1

numeros=[3, 5, 8, 12, 23, 40, 55, 70, 80]#lista tem que ser em ordem crescente
alvo=int(input("Digite o numero que você quer procurar: "))

resultado=busca_binaria(numeros,alvo)

if resultado!=-1:
   print(f"Número encontrado na posição {resultado}")
else:
    print("Número não encontrado na lista")
```
