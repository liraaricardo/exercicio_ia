20. Matriz Transposta
o Dada uma matriz (lista de listas), crie sua matriz transposta.

```python
matriz=[
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
transposta=[]

for coluna in range(len(matriz[0])):#o loop externo (coluna) percorre cada coluna da matriz original
  nova_linha=[]
  for linha in range(len(matriz)):#o loop interno (linha) percorre cada linha pegando o elemento matriz[linha]
    nova_linha.append(matriz[linha][coluna])
  transposta.append(nova_linha)

print("Matriz original:")
for linha in matriz:
    print(linha)

print("\nMatriz transposta:")
for linha in transposta:
    print(linha)
```

21. Busca Binária
o Implemente a busca binária para encontrar um elemento em uma
lista ordenada.

```python
def busca_binaria(lista,alvo):
  inicio=0
  fim=len(lista)-1 #inicio e fim marcam os limites da parte da lista que ainda estamos considerando

  while inicio<=fim:
    meio=(inicio+fim)//2 #calcula o índice do meio

    if lista[meio]==alvo: #achou,retorna a posição
      return meio

    elif lista[meio]<alvo:#o valor procurado está na metade direita, então move o inicio para depois do meio
      inicio=meio+1
    else:
      fim=meio-1
#o loop continua até inicio > fim (não sobrou mais nada pra buscar) ou até encontrar o valor,se não encontrar nada, retorna -1
  return -1

numeros=[3, 5, 8, 12, 23, 40, 55, 70, 80]#lista tem que ser em ordem crescente
alvo=int(input("Digite o numero que você quer procurar: "))

resultado=busca_binaria(numeros,alvo)

if resultado!=-1:
   print(f"Número encontrado na posição {resultado}")
else:
    print("Número não encontrado na lista")
```

22.Criptografia de Cifra de César
o Crie uma função para criptografar uma mensagem usando a cifra de
César.

```python
def cifra_cesar(texto, deslocamento):#func para criptar
    resultado = ""

    for caractere in texto:
        if caractere.isalpha():#verifica se o caractere é uma letra (ignora números, espaços, pontuação, deixando eles sem alteração)
            base = ord('A') if caractere.isupper() else ord('a')
            letra_deslocada = (ord(caractere) - base + deslocamento) % 26 #ord(caractere) converte a letra pro seu código numérico,ord(caractere) - base calcula a posição da letra no alfabeto ,+ deslocamento aplica o deslocamento desejado
            resultado += chr(base + letra_deslocada)#chr()converte número de volta pra letra,base guarda o código da primeira letra do alfabeto correspondente,chr(base + letra_deslocada) converte de volta pra letra, já deslocada
        else:
            resultado += caractere

    return resultado


mensagem = input("Digite a mensagem: ")
deslocamento = int(input("Digite o deslocamento: "))

criptografada = cifra_cesar(mensagem, deslocamento)

print(f"Mensagem criptografada: {criptografada}")
```

23. Descriptografar Cifra de César
o Crie uma função para descriptografar uma mensagem cifrada com a
cifra de César.

```python
def cifra_cesar(texto, deslocamento):
    resultado = ""

    for caractere in texto:
        if caractere.isalpha():
            base = ord('A') if caractere.isupper() else ord('a')
            letra_deslocada = (ord(caractere) - base + deslocamento) % 26
            resultado += chr(base + letra_deslocada)
        else:
            resultado += caractere

    return resultado


def decifra_cesar(texto, deslocamento):
    return cifra_cesar(texto, -deslocamento)


mensagem_cifrada = input("Digite a mensagem cifrada: ")
deslocamento = int(input("Digite o deslocamento usado na cifra: "))

original = decifra_cesar(mensagem_cifrada, deslocamento)

print(f"Mensagem original: {original}")
```

24. Validação de CPF
o Implemente uma função para validar um CPF brasileiro.

```python
def validar_cpf(cpf):
    cpf = ''.join(filter(str.isdigit, cpf))

    if len(cpf) != 11:#len(cpf) != 11 garante que sobraram exatamente 11 dígitos
        return False

    if cpf == cpf[0] * 11:#cpf == cpf[0] * 11 rejeita CPFs com todos os dígitos iguais
        return False

    soma = sum(int(cpf[i]) * (10 - i) for i in range(9))#soma tudo, multiplica por 10, tira o resto da divisão por 11
    resto = (soma * 10) % 11
    primeiro_digito = 0 if resto == 10 else resto #se o resto for 10, o dígito verificador é 0

    if primeiro_digito != int(cpf[9]):#compara esse resultado com o 10º dígito do CPF
        return False

    soma = sum(int(cpf[i]) * (11 - i) for i in range(10))#mesma lógica, mas agora usando os 10 primeiros dígitos
    resto = (soma * 10) % 11
    segundo_digito = 0 if resto == 10 else resto

    if segundo_digito != int(cpf[10]):#compara com o 11º dígito do CPF (índice 10)
        return False

    return True


cpf = input("Digite o CPF (com ou sem pontuação): ")

if validar_cpf(cpf):
    print("CPF válido")
else:
    print("CPF inválido")
```

30. Jogo da Forca
o Desenvolva um jogo simples da forca onde o usuário deve adivinhar
uma palavra.

```python
import random
palavras = ["python", "programacao", "umc", "ia", "git"]
palavra_secreta = random.choice(palavras) #random.choice(palavras) sorteia uma palavra aleatória da lista

letras_descobertas = ["_"] * len(palavra_secreta)#começa como uma lista de underscores (_), um para cada letra da palavra secreta
letras_tentadas = []#guarda as letras já usadas, pra evitar que o jogador repita a mesma letra e perca tentativa à toa
tentativas_restantes = 6

print("Bem-vindo ao Jogo da Forca!")
print(" ".join(letras_descobertas))


while tentativas_restantes > 0 and "_" in letras_descobertas:#o loop continua enquanto o jogador ainda tiver tentativas e ainda existir algum _
    letra = input("\nDigite uma letra: ").lower()

    if letra in letras_tentadas:#guarda as letras já usadas, pra evitar que o jogador repita a mesma letra e perca tentativa à toa.
        print("Você já tentou essa letra!")
        continue

    letras_tentadas.append(letra)

    if letra in palavra_secreta:#se a letra estiver na palavra, vai percorrer na palavra secreta e substituir os underscores correspondentes pela letra
        print("Boa! A letra está na palavra.")
        for i in range(len(palavra_secreta)):
            if palavra_secreta[i] == letra:
                letras_descobertas[i] = letra
    else:
        tentativas_restantes -= 1 #se não estiver, perde uma tentativa
        print(f"Letra não encontrada. Tentativas restantes: {tentativas_restantes}")

    print(" ".join(letras_descobertas))

if "_" not in letras_descobertas:
    print(f"\nParabéns! Você acertou a palavra: {palavra_secreta}")
else:
    print(f"\nVocê perdeu! A palavra era: {palavra_secreta}")
```

31. Ordenação por Bolha
o Implemente o algoritmo de ordenação por bolha (Bubble Sort).

```python
def bubble_sort(lista):
    n = len(lista)

    for i in range(n - 1):
        for j in range(n - 1 - i):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]

    return lista


numeros = [64, 25, 12, 22, 11, 90, 5]
print(f"Lista original: {numeros}")

ordenada = bubble_sort(numeros)
print(f"Lista ordenada: {ordenada}")
```

32. Jogo da Velha
o Crie um jogo da velha para dois jogadores.

```python
def imprimir_tabuleiro(tabuleiro):
    print()
    for i in range(0, 9, 3):
        print(f" {tabuleiro[i]} | {tabuleiro[i+1]} | {tabuleiro[i+2]} ")
        if i < 6:
            print("---+---+---")
    print()


def verificar_vencedor(tabuleiro):
    combinacoes = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),  # linhas
        (0, 3, 6), (1, 4, 7), (2, 5, 8),  # colunas
        (0, 4, 8), (2, 4, 6)              # diagonais
    ]

    for a, b, c in combinacoes:
        if tabuleiro[a] == tabuleiro[b] == tabuleiro[c] != " ":
            return tabuleiro[a]

    return None


tabuleiro = [" "] * 9
jogador_atual = "X"

print("Jogo da Velha!")
print("Posições do tabuleiro (digite o número correspondente):")
print(" 1 | 2 | 3 \n---+---+---\n 4 | 5 | 6 \n---+---+---\n 7 | 8 | 9 \n")

while True:
    imprimir_tabuleiro(tabuleiro)

    posicao = input(f"Jogador {jogador_atual}, escolha uma posição (1-9): ")

    if not posicao.isdigit() or int(posicao) < 1 or int(posicao) > 9:
        print("Posição inválida! Digite um número de 1 a 9.")
        continue

    posicao = int(posicao) - 1

    if tabuleiro[posicao] != " ":
        print("Posição já ocupada! Escolha outra.")
        continue

    tabuleiro[posicao] = jogador_atual

    vencedor = verificar_vencedor(tabuleiro)
    if vencedor:
        imprimir_tabuleiro(tabuleiro)
        print(f"Jogador {vencedor} venceu!")
        break

    if " " not in tabuleiro:
        imprimir_tabuleiro(tabuleiro)
        print("Empate!")
        break

    jogador_atual = "O" if jogador_atual == "X" else "X"
```

36. Gerador de Senhas
o Crie um gerador de senhas aleatórias com letras e números.

```python
import random
import string

def gerar_senha(tamanho):
    if tamanho < 4:
        return "O tamanho mínimo é 4 para garantir todos os tipos de caractere"

    minusculas = random.choice(string.ascii_lowercase)
    maiusculas = random.choice(string.ascii_uppercase)
    numeros = random.choice(string.digits)
    simbolos = random.choice("!@#$%&*")

    resto = ''.join(random.choice(string.ascii_letters + string.digits + "!@#$%&*") for _ in range(tamanho - 4))

    senha = list(minusculas + maiusculas + numeros + simbolos + resto)
    random.shuffle(senha)

    return ''.join(senha)


tamanho = int(input("Digite o tamanho da senha: "))
senha = gerar_senha(tamanho)

print(f"Senha gerada: {senha}")
```

o Implemente uma função recursiva para calcular o fatorial de um
número.

```python
def fatorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * fatorial(n - 1)


numero = int(input("Digite um número: "))
resultado = fatorial(numero)

print(f"O fatorial de {numero} é {resultado}")
```

40. Lista de Compras
o Crie um programa para gerenciar uma lista de compras, permitindo
adicionar, remover e listar itens.

```python
lista_compras = []

while True:
    print("\n--- Lista de Compras ---")
    print("1. Adicionar item")
    print("2. Remover item")
    print("3. Listar itens")
    print("4. Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        item = input("Digite o item a adicionar: ")
        lista_compras.append(item)
        print(f"'{item}' adicionado à lista.")

    elif opcao == "2":
        item = input("Digite o item a remover: ")
        if item in lista_compras:
            lista_compras.remove(item)
            print(f"'{item}' removido da lista.")
        else:
            print(f"'{item}' não está na lista.")

    elif opcao == "3":
        if lista_compras:
            print("\nItens na lista:")
            for i, item in enumerate(lista_compras, 1):
                print(f"{i}. {item}")
        else:
            print("A lista está vazia.")

    elif opcao == "4":
        print("Saindo... Até logo!")
        break

    else:
        print("Opção inválida. Tente novamente.")
```
